```python
# Crear usuarios de Argilla 

import argilla as rg

client = rg.Argilla(api_url="http://localhost:6900", api_key="argilla.apikey")

user_to_create = rg.User(
    username="usuario3",
    password="12345678",
    client=client,
)

created_user = user_to_create.create()
print(f"Usuario '{created_user.username}' creado con éxito (rol: {created_user.role}).")
```

```python
#@title Agrega usuario a workspace
import argilla as rg

client = rg.Argilla(api_url="http://localhost:6900", api_key="argilla.apikey")

user = client.users('usuario2')
#'default' es el nombre del workspace por defecto
workspace = client.workspaces('default')

added_user = user.add_to_workspace(workspace)
```
