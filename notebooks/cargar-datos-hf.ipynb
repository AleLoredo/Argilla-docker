Ese notebook permite cargar los datos del dataset publico de huggingface FineEnvs/openai-math en su correspondiente contenedor en Argilla.
Para poder hacer uso de este bloque de codigo, debe primero desde la UI de Argilla, utilizar la funcion "Importar desde HuggingFace" y crear el contenedor vacio para el dataset.

```python
import os
req_file = "../requirements.txt" if os.path.exists("../requirements.txt") else "requirements.txt"
%pip install -r {req_file}
```

```python
import argilla as rg
from datasets import load_dataset

client = rg.Argilla(
  api_url="http://localhost:6900",
  api_key="argilla.apikey"
)

ds = load_dataset("FineEnvs/openai-math", name="default", split="test")

dataset = client.datasets(name="openai-math", workspace="default")

dataset.records.log(ds, mapping={})
```
